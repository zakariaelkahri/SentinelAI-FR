# first try to build fight detection model :


```python
from ultralytics import YOLO

data_yaml = 'images/data/hazard detection.v1i.yolov8/data.yaml'
# Using YOLO v8 medium model
model = YOLO("yolov8m.pt")

results = model.train(
    data=data_yaml,
    epochs=120,
    imgsz=640,         # lower VRAM usage
    batch=8,           # fits 8GB GPU
    optimizer='SGD',
    close_mosaic=10,
    workers=0,
    device=0,
    patience=20,
    cos_lr=True,
    seed=42,
    project='runs',
    name='yolo8m_violence_detection'
)
```
