```python
from ultralytics import YOLO

data_yaml = 'images/violence2.v1i.yolov8/data.yaml'
# Using YOLO v11 medium model
model = YOLO("yolo11m.pt")

results = model.train(
    data=data_yaml,
    epochs=100,
    imgsz=640,         # lower VRAM usage
    batch=8,           # fits 8GB GPU
    optimizer='SGD',
    close_mosaic=10,
    workers=0,
    device=0,
    patience=20,
    cache=False,       # safer for large datasets
    cos_lr=True,
    seed=42,
    project='runs',
    name='yolo11m_violence_detection'
)
```
